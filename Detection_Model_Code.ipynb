```python
# Step 4e: Analyze the distribution of object classes in our dataset
def analyze_class_distribution(dataset_path, split='train'):
    """Analyze how many of each object class we have"""
    labels_dir = os.path.join(dataset_path, split, 'labels')
    label_files = glob.glob(os.path.join(labels_dir, '*.txt'))

    class_counts = {i: 0 for i in range(13)}  # We have 13 classes (0-12)

    for label_file in label_files[:500]:  # Check first 500 files for speed
        try:
            with open(label_file, 'r') as f:
                for line in f:
                    if line.strip():  # Skip empty lines
                        class_id = int(line.split()[0])
                        if class_id in class_counts:
                            class_counts[class_id] += 1
        except Exception as e:
            print(f"Error reading {label_file}: {e}")

    return class_counts

# Analyze training set
train_class_counts = analyze_class_distribution(dataset.location, 'train')

print("📈 Class Distribution in Training Set:")
print("Class ID → Class Name → Count")
for class_id, count in train_class_counts.items():
    class_name = [
        'aircraft', 'baggage_truck', 'bridge_connected', 'bus', 'catering_truck',
        'fuel_truck', 'fueling', 'ground_power', 'person', 'pushback_tractor',
        'ramp_loader', 'rolling_stairway', 'stairway'
    ][class_id]
    print(f"  {class_id:2d} → {class_name:20} → {count:4d} instances")
```

```python
# Step 4f: Visualize the class distribution
class_names = [
    'aircraft', 'baggage_truck', 'bridge_connected', 'bus', 'catering_truck',
    'fuel_truck', 'fueling', 'ground_power', 'person', 'pushback_tractor',
    'ramp_loader', 'rolling_stairway', 'stairway'
]

# Create a bar chart
plt.figure(figsize=(12, 6))
bars = plt.bar(range(13), [train_class_counts[i] for i in range(13)], color='skyblue')
plt.xlabel('Object Classes')
plt.ylabel('Number of Instances')
plt.title('Distribution of Object Classes in Training Set')
plt.xticks(range(13), class_names, rotation=45, ha='right')

# Add value labels on bars
for bar, count in zip(bars, [train_class_counts[i] for i in range(13)]):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 5, str(count),
             ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()

print("\n💡 **Key Insights:**")
print("- This shows how balanced or imbalanced our dataset is")
print("- Very common objects (like 'person') might have many examples")
print("- Rare objects might need special attention during training")
```

```python
# Step 7: Testing the trained model
from ultralytics import YOLO
import glob
import matplotlib.pyplot as plt
from PIL import Image
import os

print("🧪 Testing the trained model...")

# Load the best model from training
model_path = 'runs/detect/train/weights/best.pt'
model = YOLO(model_path)

print("✅ Best model loaded!")
print(f"Model path: {model_path}")

# Let's test on a few validation images
val_images = glob.glob(os.path.join(dataset.location, 'valid/images/*.jpg'))[:6]

print(f"Testing on {len(val_images)} sample images...")
```

```python
# Step 7b: Run predictions and display results
fig, axes = plt.subplots(2, 3, figsize=(20, 12))
axes = axes.ravel()

for i, img_path in enumerate(val_images[:6]):
    # Run prediction
    results = model.predict(img_path, conf=0.5, verbose=False)  # conf=0.5 means 50% confidence threshold

    # Plot the result
    for r in results:
        im_array = r.plot()  # plot a BGR numpy array of predictions
        im = Image.fromarray(im_array[..., ::-1])  # Convert RGB to PIL image

        axes[i].imshow(im)
        axes[i].set_title(f"Test {i+1}")
        axes[i].axis('off')

        # Print detection info
        boxes = r.boxes
        if boxes is not None:
            print(f"Image {i+1}: Detected {len(boxes)} objects")
            # Count objects by class
            class_counts = {}
            for box in boxes:
                class_id = int(box.cls[0])
                class_name = model.names[class_id]
                class_counts[class_name] = class_counts.get(class_name, 0) + 1

            # Print class distribution
            for cls, count in class_counts.items():
                print(f"   - {cls}: {count}")
        else:
            print(f"Image {i+1}: No objects detected")

plt.tight_layout()
plt.show()
```
